```python

import pandas as pd
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

users = pd.read_excel(file_name, sheet_name="Users")
courses = pd.read_excel(file_name, sheet_name="Courses")
transactions = pd.read_excel(file_name, sheet_name="Transactions")
teachers = pd.read_excel(file_name, sheet_name="Teachers")

print("All sheets loaded successfully!")
```

```python

# Step 7: Check missing values
print("===== MISSING VALUES =====")

for name, data in [
    ("Users", users),
    ("Courses", courses),
    ("Transactions", transactions)
]:
    print(f"\n{name}:")
    display(data.isnull().sum().to_frame("Missing Values"))

# Step 8: Check duplicate rows
print("\n===== DUPLICATE ROWS =====")

print("Users:", users.duplicated().sum())
print("Courses:", courses.duplicated().sum())
print("Transactions:", transactions.duplicated().sum())

# Step 9: Check data types
print("\n===== DATA TYPES =====")

print("\nUsers:")
print(users.dtypes)

print("\nCourses:")
print(courses.dtypes)

print("\nTransactions:")
print(transactions.dtypes)
```

```python

print("===== UNIQUE ID VALIDATION =====")

print("Users:")
print("Total UserIDs:", len(users))
print("Unique UserIDs:", users["UserID"].nunique())
print("Duplicate UserIDs:", users["UserID"].duplicated().sum())

print("\nCourses:")
print("Total CourseIDs:", len(courses))
print("Unique CourseIDs:", courses["CourseID"].nunique())
print("Duplicate CourseIDs:", courses["CourseID"].duplicated().sum())

print("\nTransactions:")
print("Total TransactionIDs:", len(transactions))
print("Unique TransactionIDs:", transactions["TransactionID"].nunique())
print("Duplicate TransactionIDs:",
      transactions["TransactionID"].duplicated().sum())
```

```python

print("===== AGE VALIDATION =====")

print("Minimum age:", users["Age"].min())
print("Maximum age:", users["Age"].max())
print("Age below 0:", (users["Age"] < 0).sum())
print("Age above 120:", (users["Age"] > 120).sum())

print("\n===== GENDER VALUES =====")
display(users["Gender"].value_counts(dropna=False))

print("\n===== COURSE CATEGORY VALUES =====")
display(courses["CourseCategory"].value_counts(dropna=False))

print("\n===== COURSE TYPE VALUES =====")
display(courses["CourseType"].value_counts(dropna=False))

print("\n===== COURSE LEVEL VALUES =====")
display(courses["CourseLevel"].value_counts(dropna=False))
```

```python

print("===== COURSE NUMERIC VALIDATION =====")

print("Course price below 0:",
      (courses["CoursePrice"] < 0).sum())

print("Course duration at or below 0:",
      (courses["CourseDuration"] <= 0).sum())

print("Course rating below 0:",
      (courses["CourseRating"] < 0).sum())

print("Course rating above 5:",
      (courses["CourseRating"] > 5).sum())

print("\n===== COURSE NUMERIC SUMMARY =====")
display(
    courses[
        ["CoursePrice", "CourseDuration", "CourseRating"]
    ].describe()
)
```

```python

print("===== TRANSACTION VALIDATION =====")

print("Amount below 0:", (transactions["Amount"] < 0).sum())
print("Transaction dates missing:",
      transactions["TransactionDate"].isna().sum())

print("Earliest transaction:",
      transactions["TransactionDate"].min())

print("Latest transaction:",
      transactions["TransactionDate"].max())

print("\n===== PAYMENT METHODS =====")
display(transactions["PaymentMethod"].value_counts(dropna=False))
```

```python

print("===== USER ID REFERENCES =====")

unknown_users = ~transactions["UserID"].isin(users["UserID"])
print("Transactions with unknown UserID:",
      unknown_users.sum())

print("\n===== COURSE ID REFERENCES =====")

unknown_courses = ~transactions["CourseID"].isin(courses["CourseID"])
print("Transactions with unknown CourseID:",
      unknown_courses.sum())
```

```python
print("===== TEXT FORMATTING CHECK =====")

text_columns = {
    "Users": ["UserName", "Gender"],
    "Courses": [
        "CourseName",
        "CourseCategory",
        "CourseType",
        "CourseLevel"
    ],
    "Transactions": ["PaymentMethod"]
}

datasets = {
    "Users": users,
    "Courses": courses,
    "Transactions": transactions
}

for dataset_name, columns in text_columns.items():
    print(f"\n{dataset_name} dataset")

    for column in columns:
        series = datasets[dataset_name][column].astype("string")

        extra_spaces = (
            series.notna() &
            (series != series.str.strip())
        ).sum()

        print(f"{column} - values with extra spaces: {extra_spaces}")
        print(f"{column} - unique values:")
        print(series.value_counts().to_dict())
```

```python
print("===== FINAL DATA QUALITY SUMMARY =====")

for name, df in {
    "Users": users,
    "Courses": courses,
    "Transactions": transactions
}.items():

    print(f"\n{name}")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Total missing values:", df.isna().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
# Merge transactions with user details
merged_data = pd.merge(
    transactions,
    users,
    on="UserID",
    how="left"
)

# Merge the result with course details
merged_data = pd.merge(
    merged_data,
    courses,
    on="CourseID",
    how="left"
)

print("Data integration completed successfully!")
print("Merged dataset shape:", merged_data.shape)

display(merged_data.head())
```

```python
print("===== MERGED DATA VALIDATION =====")

print("Total rows:", len(merged_data))
print("Total columns:", len(merged_data.columns))

print("\nMissing values in merged data:")
print(merged_data.isna().sum().sum())

print("\nDuplicate transaction IDs:")
print(merged_data["TransactionID"].duplicated().sum())

print("\nColumn names:")
print(merged_data.columns.tolist())
```

```python
merged_data["AgeGroup"] = pd.cut(
    merged_data["Age"],
    bins=[0, 17, 25, 35, 45, 120],
    labels=[
        "Under 18",
        "18-25",
        "26-35",
        "36-45",
        "Above 45"
    ],
    include_lowest=True
)

print("===== AGE GROUP DISTRIBUTION =====")
display(merged_data["AgeGroup"].value_counts().sort_index())
```

```python
# Convert TransactionDate to datetime format
merged_data["TransactionDate"] = pd.to_datetime(
    merged_data["TransactionDate"]
)

# Extract month and year
merged_data["EnrollmentMonth"] = (
    merged_data["TransactionDate"].dt.month
)

merged_data["EnrollmentYear"] = (
    merged_data["TransactionDate"].dt.year
)

# Extract month name
merged_data["MonthName"] = (
    merged_data["TransactionDate"].dt.strftime("%B")
)

print("===== NEW FEATURES CREATED =====")
display(
    merged_data[
        [
            "TransactionDate",
            "EnrollmentMonth",
            "EnrollmentYear",
            "MonthName",
            "AgeGroup"
        ]
    ].head()
)

print("\nUpdated dataset shape:", merged_data.shape)
```

```python
print("===== FINAL DATASET VALIDATION =====")

print("Total rows:", merged_data.shape[0])
print("Total columns:", merged_data.shape[1])

print("\nMissing values:", merged_data.isnull().sum().sum())

print(
    "Duplicate transaction IDs:",
    merged_data["TransactionID"].duplicated().sum()
)

print("\nEnrollment year distribution:")
display(merged_data["EnrollmentYear"].value_counts())

print("\nAge group distribution:")
display(merged_data["AgeGroup"].value_counts().sort_index())

print("\nFinal column names:")
print(merged_data.columns.tolist())
```

```python

import matplotlib.pyplot as plt

# Count enrollment records for each month
monthly_enrollments = (
    merged_data.groupby("EnrollmentMonth")
    .size()
    .reindex(range(1, 13), fill_value=0)
)

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

print("===== MONTHLY ENROLLMENT RECORDS =====")

monthly_table = pd.DataFrame({
    "Month": month_names,
    "EnrollmentRecords": monthly_enrollments.values
})

display(monthly_table)

# Create the chart
plt.figure(figsize=(10, 5))
plt.bar(month_names, monthly_enrollments.values, edgecolor="black")
plt.title("Monthly Course Enrollment Records - 2025")
plt.xlabel("Month")
plt.ylabel("Enrollment Records")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

print("===== MONTHLY TRANSACTION AMOUNT =====")

monthly_revenue = (
    merged_data.groupby("EnrollmentMonth")["Amount"]
    .sum()
    .reindex(range(1, 13), fill_value=0)
)

monthly_revenue_table = pd.DataFrame({
    "Month": month_names,
    "TotalAmount": monthly_revenue.values
})

display(monthly_revenue_table)

# Find highest and lowest months
print(
    "\nHighest month:",
    month_names[monthly_revenue.values.argmax()],
    "-",
    round(monthly_revenue.max(), 2)
)

print(
    "Lowest month:",
    month_names[monthly_revenue.values.argmin()],
    "-",
    round(monthly_revenue.min(), 2)
)

# Plot monthly transaction amounts
plt.figure(figsize=(10, 5))
plt.plot(
    month_names,
    monthly_revenue.values,
    marker="o"
)
plt.title("Monthly Transaction Amount - 2025")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

print("===== AVERAGE COURSE RATING BY CATEGORY =====")

# Each course should count once in this calculation
category_ratings = (
    courses.groupby("CourseCategory")["CourseRating"]
    .mean()
    .sort_values(ascending=False)
)

display(category_ratings.round(2))

# Visualize average ratings
plt.figure(figsize=(10, 6))

category_ratings.sort_values().plot(
    kind="barh",
    edgecolor="black"
)

plt.title("Average Course Rating by Category")
plt.xlabel("Average Rating")
plt.ylabel("Course Category")
plt.xlim(0, 5)
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Count enrollments for each course
course_enrollment_counts = (
    merged_data["CourseID"].value_counts()
    .rename("EnrollmentCount")
)

# Create one row per course, with its price and enrollment count
course_analysis = courses.merge(
    course_enrollment_counts,
    left_on="CourseID",
    right_index=True,
    how="left"
)

course_analysis["EnrollmentCount"] = (
    course_analysis["EnrollmentCount"].fillna(0).astype(int)
)

print("===== COURSE PRICE AND ENROLLMENTS =====")

display(
    course_analysis[
        ["CourseName", "CoursePrice", "EnrollmentCount"]
    ].sort_values(
        "EnrollmentCount", ascending=False
    ).head(10)
)

# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(
    course_analysis["CoursePrice"],
    course_analysis["EnrollmentCount"]
)
plt.title("Course Price vs Enrollment Count")
plt.xlabel("Course Price")
plt.ylabel("Enrollment Records")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Calculate correlation
correlation = course_analysis[
    ["CoursePrice", "EnrollmentCount"]
].corr().iloc[0, 1]

print("\nPrice-enrollment correlation:", round(correlation, 3))
```
